Key Metrics

```python
print('\n1. Customer level Metric' )
customer_metric=orders.groupby('customer_id').agg({'order_id':'count',
                                                   'amount':['sum','mean'],
                                                   'order_date':'max'}).reset_index()

customer_metric.columns=['customer_id', 'num_orders', 'total_spent', 'avg_order_value', 'last_order_date']
```

```python
customer_metric
```

```python
# Calculate days since last order (recency)

today=orders['order_date'].max()
customer_metric['days_since_last_order']=(today-customer_metric['last_order_date']).dt.days
customer_metric
```

```python
print("\n1. Customer purchase behavior:")
print(f"   Average # of orders per customer: {customer_metric['num_orders'].mean():.1f}")
print(f"   Average total spent per customer: ${customer_metric['total_spent'].mean():,.2f}")
print(f"   Max orders by single customer: {customer_metric['num_orders'].max()}")
print(f"   Max spent by single customer: ${customer_metric['total_spent'].max():,.2f}")
```

```python
print("\n2️. Purchase recency (days since last order):")
print(f"   Average: {customer_metric['days_since_last_order'].mean():.1f} days")
print(f"   Median: {customer_metric['days_since_last_order'].median():.1f} days")
print(f"   Min (most recent): {customer_metric['days_since_last_order'].min()} days")
print(f"   Max (most inactive): {customer_metric['days_since_last_order'].max()} days")
```

```python
print("\n3️. Preview of customer metrics:")
print(customer_metric.head(10))
```

```python
print('\n4. Segment Analysis')
#Merge with customer_segments
customer_metric_with_segment= customer_metric.merge(customers[['customer_id', 'segment', 'region']], on='customer_id')
customer_metric_with_segment
```

```python
segment_summary=customer_metric_with_segment.groupby('segment').agg({'num_orders':['sum','mean'],
                                                                     'total_spent':['count','mean','sum',],
                                                                     'days_since_last_order': 'mean'

                                                                     }).round(2)
```

```python
print("\nAverage metrics by segment:")
print(f"{'Segment':<15} {'Count':<10} {'Avg Spent':<15} {'Total $':<15} {'Avg Orders':<12} {'Days Inactive':<12}")
print("-" * 80)
for segment in ['Standard', 'Premium', 'Enterprise']:
    data = customer_metric_with_segment[customer_metric_with_segment['segment'] == segment]
    print(f"{segment:<15} {len(data):<10} ${data['total_spent'].mean():<14,.0f} ${data['total_spent'].sum():<14,.0f} {data['num_orders'].mean():<12.1f} {data['days_since_last_order'].mean():<12.1f}")
```

## Customer Churn Analysis

```python
# Defining churn risk based on recency
# If no order in last 60 days = HIGH RISK

customer_metric['churn risk']=customer_metric['days_since_last_order']>=60

print(f"\n1️. At-risk customers (no order in 60+ days):")
at_risk_count=customer_metric['churn risk'].sum()
print(f"   Count: {at_risk_count} customers")
print(f"   % of customers: {(at_risk_count/len(customer_metric)*100):.1f}%")
```

```python
print(f"\n2️. Risk distribution:")
print(f"   Active (0-30 days): {(customer_metric['days_since_last_order'] < 30).sum()}")
print(f"   Moderate (30-60 days): {((customer_metric['days_since_last_order'] >= 30) & (customer_metric['days_since_last_order'] < 60)).sum()}")
print(f"   At Risk (60+ days): {(customer_metric['days_since_last_order'] >= 60).sum()}")
```

```python
print(f"\n3️. At-risk customers by segment:")
at_risk_by_segment = customer_metric_with_segment.copy()
at_risk_by_segment['churn_risk'] = at_risk_by_segment['days_since_last_order'] >= 60


for segment in ['Standard', 'Premium', 'Enterprise']:
    segment_data = at_risk_by_segment[at_risk_by_segment['segment'] == segment]
    at_risk = segment_data['churn_risk'].sum()
    total = len(segment_data)
    print(f"   {segment}: {at_risk}/{total} at risk ({(at_risk/total*100):.1f}%)")
```

```python
print(f"\n4️.  Top 10 at-risk customers (no orders in 60+ days):")

at_risk_customers = customer_metric_with_segment[customer_metric_with_segment['days_since_last_order'] >= 60].sort_values('days_since_last_order', ascending=False).head(10)
print(at_risk_customers[['customer_id', 'segment', 'total_spent', 'num_orders', 'days_since_last_order']])
```
